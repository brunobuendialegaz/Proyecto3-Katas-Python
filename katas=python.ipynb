1.- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
cadena_texto = input('Introduce el texto:')

def contar_caracteres(texto):
    texto_lower = texto.lower()
    dic_letras = {}
    for letra in texto_lower:
        if letra == ' ':
            continue
        elif letra not in dic_letras:
            dic_letras[letra] = 1
        else:
            dic_letras[letra] += 1
    return dic_letras

print(contar_caracteres(cadena_texto))
```

5.- Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
notas = [5,7,3,5,9,10]

def calcular_aprobado(listado_notas, nota_corte=5):
    nota_total = 0
    for item in listado_notas:
        nota_total += item
    media = nota_total / len(listado_notas)
    return (media, "aprobado" if media >= nota_corte else "suspenso")

print(calcular_aprobado(notas))
```

6.- Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def calcular_factorial(num, total=1):
    if num <= 1:
        return total
    else:
        total *= num
        num -= 1
        return calcular_factorial(num, total)

print(calcular_factorial(4))
```

9.- Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().

```python
def filtrar_mascotas(lista_mascotas):
    prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]
    return list(filter(lambda mascota: mascota not in prohibidas, lista_mascotas))

mascotas = ["Perro", "Tigre", "Gato", "Mapache", "Canario", "Oso", "Conejo"]
mascotas_permitidas = filtrar_mascotas(mascotas)

print(mascotas_permitidas)
```

10.- Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.

```python
def calcular_promedio(*args):
    total = 0
    for item in args:
        total += item
    try:
        return total / len(args)
    except ZeroDivisionError:
        return "La lista esta vacía"

print(calcular_promedio(5,8,9,5,2,1,5,6))
```

12.- Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().

```python
frase_input = input("Introduce la frase a medir")

def medidor_longitud_palabras(frase):
    lista_palabras = frase.split()
    lista_longitudes = list(map(len, lista_palabras))
    for indice, palabra in enumerate(lista_palabras):
        print(f"Palabra -> {palabra}, longitud -> {lista_longitudes[indice]}")
    return lista_longitudes

print(medidor_longitud_palabras(frase_input))
```

17.- Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().

```python
from functools import reduce

listado_numeros = [5,7,2]

def concatenar_numeros(lista):
    return int(reduce(lambda acumulador, d: acumulador * 10 + d, lista))

print(concatenar_numeros(listado_numeros))
```
