1.- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
cadena_texto = input('Introduce el texto:')

def contar_caracteres(texto):
    texto_lower = texto.lower()
    dic_letras = {}
    for letra in texto_lower:
        if letra == ' ':
            continue
        elif letra not in dic_letras:
            dic_letras[letra] = 1
        else:
            dic_letras[letra] += 1
    return dic_letras

print(contar_caracteres(cadena_texto))
```

```python
notas = [5,7,3,5,9,10]

def calcular_aprobado(listado_notas, nota_corte=5):
    nota_total = 0
    for item in listado_notas:
        nota_total += item
    media = nota_total / len(listado_notas)
    return (media, "aprobado" if media >= nota_corte else "suspenso")

print(calcular_aprobado(notas))
```

6.- Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def calcular_factorial(num, total=1):
    if num <= 1:
        return total
    else:
        total *= num
        num -= 1
        return calcular_factorial(num, total)

print(calcular_factorial(4))
```

30.- Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.


```python
palabra = "amor"
palabra_anagrama = "roma"
palabra_random = "barco"

def comprobar_anagrama(p1, p2):
    chars1 = []
    chars2 = []
    acc = 0
    if len(p1) != len(p2):
        return False
    while acc < len(p1):
        chars1.append(p1[acc])
        chars2.append(p2[acc])
        acc += 1
    chars1.sort()
    chars2.sort()
    return chars1 == chars2

print(comprobar_anagrama(palabra, palabra_anagrama))
print(comprobar_anagrama(palabra, palabra_random))
```

34.- Crea la clase Arbol
Define un árbol genérico con un tronco y ramas como atributos.

Métodos disponibles: crecer_tronco, nueva_rama, crecer_ramas, quitar_rama, info_arbol.

Código a seguir:

Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.

Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.

Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.

Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.

Implementar el método quitar_rama para eliminar una rama en una posición específica.

Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.

Caso de uso:

        a. Crear un árbol.

        b. Hacer crecer el tronco una unidad.

        c. Añadir una nueva rama.

        d. Hacer crecer todas las ramas una unidad.

        e. Añadir dos nuevas ramas.

        f. Retirar la rama situada en la posición 2.

        g. Obtener información sobre el árbol.


```python
class Arbol:
    def __init__(self):
        self.tronco = 1
        self.ramas = []

    def crecer_tronco(self):
        self.tronco += 1

    def nueva_rama(self):
        self.ramas.append(1)

    def crecer_ramas(self):
        for rama in self.ramas:
            rama += 1

    def quitar_rama(self, posicion):
        if posicion >= 0 and posicion < len(self.ramas):
            self.ramas.pop(posicion)
        else:
            print(f"La rama de la posición {posicion} no existe")

    def info_arbol(self):
        print(f"La longitud del arbol es {self.tronco}")
        print(f"Estas son sus ramas: {self.ramas}")

mi_arbol = Arbol()

mi_arbol.crecer_tronco()

mi_arbol.nueva_rama()

mi_arbol.crecer_ramas()

mi_arbol.nueva_rama()
mi_arbol.nueva_rama()

mi_arbol.quitar_rama(2)

mi_arbol.info_arbol()
```

Código a seguir:

Inicializar un usuario con nombre, saldo y un indicador (True o False) de cuenta corriente.

Implementar retirar_dinero para sustraer dinero del saldo, lanzando un error si no es posible.

Implementar agregar_dinero para aumentar el saldo del usuario.

Caso de uso:

        a. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
        b. Agregar 20 unidades al saldo de Bob.
        c. Transferir 80 unidades de Bob a Alicia.
        d. Retirar 50 unidades del saldo de Alicia.

```python
class UsuarioBanco:
    def __init__(self, nombre = None, saldo = 0.0, cuenta_corriente = False):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero(self, cantidad):
        if cantidad <= self.saldo:
            self.saldo -= cantidad
            return f"Cantidad {cantidad} retirada, saldo disponible: {self.saldo}"
        else:
            return f"Cantidad insuficiente, saldo actual: {self.saldo}"

    def agregar_dinero(self, cantidad):
        if cantidad > 0:
            self.saldo += cantidad
            return f"Cantidad añadida, saldo actual: {self.saldo}"

    def transferir_dinero(self, usuario_banco, cantidad):
        if self.saldo >= cantidad and isinstance(usuario_banco, UsuarioBanco):
            self.retirar_dinero(cantidad)
            usuario_banco.agregar_dinero(cantidad)
            return f"Cantidad transferida, saldo actual: {self.saldo}"
        return f"Saldo insuficiente. Saldo actual: {self.saldo}"

alicia_cuenta = UsuarioBanco("Alicia", 100, True)
bob_cuenta = UsuarioBanco("Bob", 50, True)

print(bob_cuenta.agregar_dinero(20))

print(bob_cuenta.transferir_dinero(alicia_cuenta, 80))

print(alicia_cuenta.retirar_dinero(50))


```

```python
def calificador_notas(nota):
    if nota >= 0 and nota < 70:
        return "Insuficiente"
    elif nota >= 70 and nota < 80:
        return "bien"
    elif nota >=80 and nota < 90:
        return "Muy bien"
    elif nota >= 90 and nota <= 100:
        return "Excelente"
    else:
        return "Nota no contemplada (0-100)"

print(calificador_notas(15))
print(calificador_notas(-15))
print(calificador_notas(72))
print(calificador_notas(88))
print(calificador_notas(100))
```

39.- Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

```python
import math

def calcular_area(figura, datos):
    if figura == "rectangulo":
        return datos[0] * datos[1]
    elif figura == "circulo":
        return math.pi * (datos ** 2)
    elif figura == "triangulo":
        return (datos[0] * datos[1]) / 2
    else:
        return "La figura no esta contemplada, prueba rectangulo, circulo o triangulo"

print(calcular_area("rectangulo", (5, 10)))
print(calcular_area("circulo", 3))
print(calcular_area("triangulo", (5, 10)))
```
