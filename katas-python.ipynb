1.- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
cadena_texto = input('Introduce el texto:')

def contar_caracteres(texto):
    texto_lower = texto.lower()
    dic_letras = {}
    for letra in texto_lower:
        if letra == ' ':
            continue
        elif letra not in dic_letras:
            dic_letras[letra] = 1
        else:
            dic_letras[letra] += 1
    return dic_letras

print(contar_caracteres(cadena_texto))
```

5.- Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
notas = [5,7,3,5,9,10]

def calcular_aprobado(listado_notas, nota_corte=5):
    nota_total = 0
    for item in listado_notas:
        nota_total += item
    media = nota_total / len(listado_notas)
    return (media, "aprobado" if media >= nota_corte else "suspenso")

print(calcular_aprobado(notas))
```

6.- Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def calcular_factorial(num, total=1):
    if num <= 1:
        return total
    else:
        total *= num
        num -= 1
        return calcular_factorial(num, total)

print(calcular_factorial(4))
```

30.- Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.


```python
palabra = "amor"
palabra_anagrama = "roma"
palabra_random = "barco"

def comprobar_anagrama(p1, p2):
    chars1 = []
    chars2 = []
    acc = 0
    if len(p1) != len(p2):
        return False
    while acc < len(p1):
        chars1.append(p1[acc])
        chars2.append(p2[acc])
        acc += 1
    chars1.sort()
    chars2.sort()
    return chars1 == chars2

print(comprobar_anagrama(palabra, palabra_anagrama))
print(comprobar_anagrama(palabra, palabra_random))
```

```python
def pedir_nombres():
    lista_nombres = [] 
    try: # aqui he intentado meter un poco de control de errores, como en ej anteriores
        numero = int(int(input("introduce el numero de nombres a añadir")))
        if numero < 1 or numero > 10:
            raise ValueError(f"el numero {numero} no esta en el rango permitido (1-10)")
    except ValueError as e: 
            return f"Error {e}"
    else:
        acc = 1
        while acc <= numero:
            lista_nombres.append(input(f"Introduce el nombre {acc}"))
            acc += 1
        return lista_nombres


    
    
def buscador_nombres():
    lista_nombres = pedir_nombres()
    nombre = input("Introduce un nombre para buscar")
    for item in lista_nombres:
        if item == nombre:
            return f"Nombre encontrado: {nombre}"               
    raise ValueError(f"El nombre {nombre} no esta en la lista: {lista_nombres}")

buscador_nombres()    



```
